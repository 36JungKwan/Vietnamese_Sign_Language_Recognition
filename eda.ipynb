```python
import numpy as np

# Thay đường dẫn này bằng đường dẫn thực tế tới 1 file .npz bất kỳ trên máy bạn
sample_path = r"D:/VSL_data/Processed/train/1/1.npz"

try:
    data = np.load(sample_path, allow_pickle=True)
    print("1. Các thành phần (keys) có trong file .npz:", data.files)
    
    print("\n2. Kích thước (Shape) của từng thành phần:")
    for key in data.files:
        print(f" - {key}: {data[key].shape}")
        
except Exception as e:
    print(f"Lỗi khi đọc file: {e}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Thay đường dẫn tới file .npz trên máy bạn
file_path = r"D:/VSL_data/Processed/train/1/1.npz"
data = np.load(file_path)
seq = data['sequence'] # Shape: (60, 201)

# Lấy frame đầu tiên (frame 0)
frame_0 = seq[0] 

# GIẢ ĐỊNH 1: Dữ liệu được lưu dạng Interleaved [x0, y0, z0, x1, y1, z1...]
try:
    interleaved_nodes = frame_0.reshape(67, 3)
    x = interleaved_nodes[:, 0]
    y = interleaved_nodes[:, 1]
    
    plt.figure(figsize=(6, 6))
    plt.scatter(x, y, c='blue', s=20)
    plt.title("Phân bố 67 điểm ảnh (Giả định Interleaved)")
    plt.gca().invert_yaxis() # Lật trục Y vì tọa độ ảnh thường có gốc ở góc trái trên
    plt.show()
except Exception as e:
    print("Lỗi khi test Interleaved:", e)
```
